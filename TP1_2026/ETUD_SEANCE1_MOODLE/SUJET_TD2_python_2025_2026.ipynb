### Les instructions conditionnelles
    2) ALTERNATIVE ou CHOIX : Exécution d'une séquence d'actions si la condition est vérifiée,               
               exécution d'une autre séquence si la condition n'est pas vérifiée
        
           SI condition 
               ALORS bloc_action_1 # si la valeur de la condition est Vrai
               SINON bloc_action_2 # si la valeur de la condition est Faux
           
           Code Python # en python les indentations déterminent le bloc
           if condition1 :
               bloc1 # si la valeur de la condition est Vrai
           else :
               bloc2 # si la valeur de la condition est Faux

### Les instructions conditionnelles
    3) CHOIX IMBRIQUÉS : Exécution d'une séquence d'actions (et donc du code) si la première condition est 
       vérifiée sinon on tec-ste une deuxième condition. Si celle-ci est vérifiée oon exécute la séquence 
       correspondante sinon ...
       
       Imbrication de SI .. ALORS .. SINON 
              
           SI condition1
               ALORS bloc_action_1      # si la valeur de la condition1 est Vrai
               SINON SI condition 2     # si la valeur de la condition1 est Faux
                   ALORS bloc_action_2  # si la valeur de la condition 2 est Vrai
                   SINON SI condition 3 # si la valeur de la condition 2 est Faux 
                       ALORS ...
           
           Code Python # en python les indentations déterminent le bloc
           if condition1 :
               bloc1 # si la valeur de la condition1 est Vrai
           elif condition2 : 
               bloc2 # si la valeur de la condition1 est Faux et que la condition2 est à Vrai
           elif condition3 :
               bloc3 # si les valeurs des condition 1 et condition2 sont à Faux et condition3 à Vrai
               ...
           else :    # si aucune des condition précédentes n'est à Vrai
               blocN

### Les instructions conditionnelles
    4) Combinaisons de conditions
    
       une condition peut être composée de variables booléennes, d'expressions relationnelles ou d'opérations logiques 
       booléens (et, ou, non) entre conditions plus élémentaires. 
       
           SI condition1 et condition 2 ALORS ...
           SI condition 1 ou (condition 2 et non condition3) ALORS ...
           
       Code Python # en python les indentations déterminent le bloc
       if condition1 and condition 2 :
           bloc1 
           # si la valeur de la condition1 ET de la condition 2 sont Vrai
       elif condition 3 or condition4 : 
           bloc2 
           # (condition1 OU condition 2 sont à Faux) ET (condition3 OU condition 4 sont à Vrai)
       elif non condition5
           bloc3 
           # (condition 1 OU condition2 à Faux) ET (condition 3 ET condition 4 à Faux) et condition5 à Faux
           ...
       else :    
       # au final (condition 1 OU condition2 à Faux) ET (condition 3 ET condition 4 à Faux) et condition5 à Vrai
           blocN

### Les instructions répétitives / itératives
        
    1) Boucle POUR : utilisable si le nombre de répétitions (passage dans la boucle) est connu à l'avance. Nécessite 
    l'utilisation d'une variable de boucle (entier utilisé comme un compteur de tours) et un intervalle de parcours : 
    valeur de début et valeur de fin. Le pas par défaut est de 1 mais peut être modifié explicitement
       
           POUR N de 1 à 10 FAIRE  # N prendra tour à tour les valeurs 1, 2, 3, ... 10
               action ; 
           FIN POUR
           
           POUR N de 2 à 10 PAS 2 FAIRE # N prendra tour à tour les valeurs 2, 4, 6, 8, 10
               actions ;
           FIN POUR
           
           Code Python # en python les indentations déterminent le bloc
           for n in intervalle : 
               bloc_instruction_a_repeter
