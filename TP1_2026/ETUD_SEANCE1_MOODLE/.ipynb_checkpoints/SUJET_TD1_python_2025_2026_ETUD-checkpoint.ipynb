## TD1 : Algorithmes de base et premiers programmes Python


### Typage dynamique ... quesaco ? 
      

### Typage dynamique ... quesaco ? 
        1) Association d'une variable à un type de données
            - fait de manière explicite et une fois pour toute (langage C) = NON modifiable 
            - fait de manière implicite au moment de son utilisation (python) = ATTENTION modifiable
     

### Typage dynamique ... quesaco ? 
        
        2) Conversion possible de la valeur d'une variable mors de son utilisation = CAST
            - considère une valeur (variable, expression, résultat d'une fonction, ...) 
            - la convertit dans le type indiqué 
                --> ne modifie pas le type de la variable 
                --> possible sous réserve de compatibilité de type
                --> possible en C, en python, ...syntaxe différente mais même effet
        

### Typage dynamique ... quesaco ? 
        
     
        3) EN PYTHON : typage dynamique = fonction de la valeur attribuée à la variable ou à la constante
        - type(variable) : permet de connaître le type de la variable
        - id(variable) : permet de connaître l'identifiant d'une variable 
                         
##### ATTENTION : deux variables ayant des noms différents mais le même identifiant représentent la même donnée en mémoire !!!
        
       

Types de données de base 
    - nombre entier = integer
    - nombre flottant (virgule flottante, décimal, réel) = float
    - chaine de carcatères = string
    - booléen = boolean
    
Typage dynamique ==> SIMPLE et PRATIQUE 
MAIS ... modifiable en cours d'exécution = ATTENTION DANGER !!

```python
# EXEMPLE 2
nb_entier = 3
print(type(nb_entier))

nb_flottant = 3.5
print(type(nb_flottant))

chaine = "de caractères"
print(type(chaine), id(chaine))

car = 'x'
print(type(car))

cond_ok = True
print(type(cond_ok))

# ATTENTION : la donnée précédente de même nom est perdue !!!!
chaine = 10.25
print(type(chaine), id(chaine))
```

#### QUESTION B 
Exécuter le code suivant

Observer le résultat et commenter
Quel est l'effet du typage dynamique ici ?
Que se passe-t-il pour la variable nb ?

```python
nb = int(3)
print(nb, '\t', type(nb), id(nb))

nb = float(3)
print(nb, '\t', type(nb), id(nb))

nb = str(3)
print(nb, '\t', type(nb), id(nb))

print ('-------------------')
    
nb = int(3.5)
print(nb, '\t', type(nb), id(nb))

nb = float(3.5)
print(nb, '\t', type(nb), id(nb))

nb = str(3.5)
print(nb, '\t', type(nb), id(nb))

print ('-------------------')

nb = bool(0)
print(nb, '\t', type(nb), id(nb))

nb = bool(1)
print(nb, '\t', type(nb), id(nb))

nb = bool(3.5)
print(nb, '\t', type(nb), id(nb))

print ('-------------------')

nb = int('a')
print(nb, '\t', type(nb), id(nb))
```

#### QUESTION F
- Etendre ce programme au calcul de la surface d'un cercle et du volume d'une sphère de même rayon
- Améliorer la lisibilité du programme et l'interaction avec l'utilisateur
- Proposer une version arrondie des valeurs en appliquant la fonction round(x,nb) la valeur de x est alors arrondie à nb décimales. 


```python
# EXEMPLE 1.4
from math import pi
print("Entrée du rayon voulu : ")
#r1 = float(input())
r1 = float(2)
r2 = 2*pi*r1
surfaceCercle = pi*r1**2
surfaceSphere = 4*pi*r1**2
surfaceCercleArrondie = round(surfaceCercle,0)
surfaceSphereArrondie = round(surfaceSphere,0)
print("Affichage du rayon puis du diametre")
print(r1,r2)  

print("La surface du cercle est : ", surfaceCercle)
print("La surface du cercle arrondie est : ", surfaceCercleArrondie)
print("La surface de la sphere est : ", surfaceSphere)
print("La surface de la sphere arrondie est : ", surfaceSphereArrondie)
```
