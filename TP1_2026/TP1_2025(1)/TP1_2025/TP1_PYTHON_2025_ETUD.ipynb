### EXERCICE 1 : Instructions conditionnelles

Un pixel est un constituant de base d'une image. Il correspond à une valeur entière qui représente une intensité. Les valeurs des pixels sont comprises entre 0 et 255 (256 valeurs possibles). 

Dans une image noir et blanc on considère que 1 correspond au BLANC et 0 au NOIR. 
Dans une image en niveaux de gris on considère toute la palette des gris en allant du noir (0) au blanc (255) en passant par tous les degrés de gris possible du plus foncé au plus clair. 


#### Exercice 1.1 :

- valeur non comprise dans cet intervalle --> message 'valeur non valide'

```python
# Exercice 1.1 
#Lire une valeur de pixel

#Variables

#Debut
pixel = int(input('Entrez une valeur de pixel entre 0 et 255 : ')) #Entrée d'une valeur de pixel par l'utilisateur
#print(pixel)

#Test de la valeur du pixel pour validation
if pixel < 0 or pixel > 255:
    print("Valeur non valide")
else :
    print("Valeur valide")
    print("Votre pixel vaut : ", pixel)
#FIN
```

#### Exercice 1.5 : en partant de la copie du programme précédent modifier le code pour obtenir le traitement suivant :  

généraliser le traitement pour ramener la valeur du pixel à un autre nombre de niveaux de gris (nb_niveaux) en calculant la valeur pixel_Nb et en affichant le message  'valeur sur nb_niveaux de gris = pixel_Nb'
dans la mesure ou nb_niveaux est : 
- (1) lu au clavier 
- (2) doit être compris entre 1 et 255 
- (3) correspondre à une puissance de 2 

Le programme devra vérifier la validité de nb_niveaux avant de procéder à la lecture et au traitement du pixel. 

Si la valeur n'est pas valide le message 'la valeur fournie pour nb_niveaux n'est pas valide' sera affiché (ainsi que la raison : soit en dehors de l'intervalle ou bien dans l'intervalle mais pas une puissance de 2. 

En cas de validité de nb_niveaux, le traitement sera effectué et le message prévu affiché

Vous utiliserez pour cela la bibliothèque math que vous aurez importée au préalable avec 

    import math

sachant que la fonction log2(x) de cette bibliothèque doit vous aider à faire le test puissance de 2 ou pas 

Effectuer les tests sur les valeurs suivantes et copier-coller le rapport de tests ci-dessous : 300, 15, 16, 8, 0, ...


```python
# Exercice 1.5
#Lire une valeur de pixel
import math

#Variables

#Debut
nb_niveaux = int(input('Entrez le nombre de niveau de gris souhaité, celui doit correspondre à une puissance de 2 entre 1 et 255 : '))

if (nb_niveaux > 0 and nb_niveaux < 256):
    puissance = math.log2(nb_niveaux)
    if (puissance != int(puissance)):
        print(nb_niveaux, "n'est une puissance de 2")
    else:
        limite = 256/nb_niveaux # calcul de la limite 
        pixel = int(input('Entrez une valeur de pixel entre 0 et 255 : ')) #Entrée d'une valeur de pixel par l'utilisateur
        #print(pixel)

        #Test de la valeur du pixel pour validation
        if pixel < 0:
            print("La valeur fournie doit être positive")
        elif pixel > 255:
            print("La valeur fournie est trop grande")
        else:
            print("La valeur fournie est correcte")
            pixel_n = pixel/limite #calcul du niveau du pixel
            print("Votre pixel à un niveau de gris de : ", int(pixel_n))      
else:
    print("La valeur fournie pour nb_niveaux n'est pas valide, vous êtes en dehors de l'interval")

#FIN
```

### Exercice 3 : Structures de données, traitements itératifs et gestion de tableaux à deux dimensions

#### Exercice 3.1 : 

Une image de  16 x 16 pixels peut être créée en recopiant automatiquement 16 fois la valeur du tableau initial. On a ainsi une liste de listes que l'on peut interpréter comme une image en niveau de gris. Pour cela, exécuter le code suivant et étudiez-le. 

```python
import matplotlib as mp
from matplotlib import pyplot
tab = []
print(tab)
long_tab = len(tab)
mon_image = []

# construction de l'image à partir de copies de la ligne tab
for i in range(long_tab):
    mon_image.append(tab.copy())
    # vérification des identifiants --> ils doivent être différents 
    print('-----', id(mon_image[i]))
    
print('liste de liste ', type(mon_image), '\n',type(mon_image[0]), '\n', mon_image)
# visualisation de l'image en niveau de gris
mp.pyplot.imshow(mon_image, cmap='gray')
```
