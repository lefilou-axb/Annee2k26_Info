```python
# Exercice 1.2 - Image en négatif

# bibliothèque à importer

import numpy as np                        # pour la gestion des tableaux de type array
import matplotlib as mpl                  # pour travailler avec tout type de figures
import matplotlib.pyplot as plt           # pour tracer des figures      

# mémorisation du chemin vers le dossier contenant les données à traiter
PATH = '/home/hgl3968a/Annee3_2/Informatique/TP2_2026/ETUD_TP2_SUJETS'

# chargement d'une image donnée noir et blanc - résultant d'une binarisation
# utilisation des fonctionnalités de la bibliothèque MATPLOTLIB
# résultat directement en np.array

mon_image_nb = mpl.image.imread(PATH + '/TEST_TP_PYTHON/56.bmp')

# calcul des valeurs des dimensions de l'image lue
largeur, hauteur = mon_image_nb.shape

image_Neg = mon_image_nb.copy()

for i in range(largeur):
    for j in range(hauteur):
        if image_Neg[i][j] == 0 : 
            image_Neg[i][j] = 255
        else :
            image_Neg[i][j] = 0

# affichage de la matrice directement dans la trace
print(type(image_Neg), '\n')    
print(image_Neg)


# Valeurs des 2 dimensions récupérées
print('Taille de l\'image = ', hauteur, ' x ', largeur, '\n')  

# type du tableau et des éléments du tableau
print('type élément = ', type(image_Neg[0][0]), '\n')

# affichage de la matrice comme une image en niveaux de grix
plt.imshow(image_Neg, cmap='gray')
```
